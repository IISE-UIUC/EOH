```python
# imports

# "cv2" is the Python library that does Computer Vision
import cv2
import numpy as np
```

```python
# create image variable from .jpg using "imread"
image = cv2.imread('test_image.jpg')
    # in SCM, we will use a new image from a video feed for each frame

# images are given height, width, and number of channels
print(image.shape)
print()

# images are numpy arrays
print(image[150, 300, :])
```

```python
# save an image (in this case, saving the image we just took with a new name)
cv2.imwrite('test_image_2.jpg', image)

# display an image in a separate window
cv2.imshow('window name', image)

# waits for a key press for a certain number of milliseconds
cv2.waitKey(0)
    # pauses the program for however long is stated
    # if 0, it waits indefinately until the window is closes

# closes all windows, the window breaks and Python gets mad if it's not here
cv2.destroyAllWindows()
```

```python
# read a camera feed
webcam = cv2.VideoCapture(0)
    # the "0" refers to the camera you are using
    # if you have multiple cameras, the one you want may be on "1", "2", etc.

# display webcam
while True:
    # read webcam feed: "frame" is the image and "ret" is a boolean that is True if the webcam is working
    ret, frame = webcam.read()
    # display
    cv2.imshow('window name', frame)
    # program waits 40 milliseconds before looping again (waitKey controls framerate)
    # if the "spacebar" is pressed, break the loop
        # ord(' ') refers to the spacebar key, but you can also use 32, the ASCII code for spacebar
    if cv2.waitKey(40) & 0xFF == ord(' '):
        # "break" ends the while loop, so the program moves on and closes the window
        break

webcam.release()
cv2.destroyAllWindows()
```

```python
# Resizing / Cropping an image

image = cv2.imread('test_image.jpg')
print(image.shape)

resized_image = cv2.resize(image, (1280, 400), interpolation=cv2.INTER_AREA)
cropped_image = image[50:250, 100:500]

cv2.imshow('resized', resized_image)
cv2.imshow('cropped', cropped_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Color Spaces

image = cv2.imread('test_image.jpg')

# convert image between color spaces
# HSV color space: "Hue Saturation Value"
    # used for deciding the range of color to detect
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV) # converting to HSV color space
# grayscale color space
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imshow('BGR color space (normal)', image) # display normal image
cv2.imshow('HSV color space', image_hsv) # display image in HSV color space
cv2.imshow('GRAY color space', image_gray)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Blurring

image = cv2.imread('test_image.jpg')

# blur an image by avaraging the nearby pixels
    # in this case, averages value of 5 pixels to the right and left, and 5 above and below
image_blur = cv2.blur(image, (5, 5))

# different type of blur
image_GausBlur = cv2.GaussianBlur(image, (5, 5), 3)

cv2.imshow('normal', image)
cv2.imshow('blur()', image_blur)
cv2.imshow('GaussianBlur()', image_GausBlur)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Threshold: makes a binary image, pushes all color values to 0 or 255
    # used for separating images from their background

image = cv2.imread('test_image.jpg')
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) # convert to grayscale

# X is the threshold value
    # all color values below 80 go to 0
    # all color values above 80 go to 255
X = 80
# create the binary image using a (simple) threshold
ret, image_thresh = cv2.threshold(image_gray, X, 255, cv2.THRESH_BINARY)

cv2.imshow('grayscale', image_gray)
cv2.imshow('threshold', image_thresh)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Threshold with Blur

image_thresh = cv2.blur(image_thresh, (5, 5))
ret, image_thresh = cv2.threshold(image_thresh, X, 255, cv2.THRESH_BINARY)

cv2.imshow('grayscale', image_gray)
cv2.imshow('threshold', image_thresh)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Adaptive Threshold: when a simple threshold doesn't isolate the right things

image = cv2.imread('thresh_image.jpg')
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) # convert to grayscale

# regular threshold, threshold value is 60
ret, image_thresh = cv2.threshold(image_gray, 60, 255, cv2.THRESH_BINARY)
# adaptive threshold
image_adapt = cv2.adaptiveThreshold(image_gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 21, 30)

cv2.imshow('grayscale', image_gray)
cv2.imshow('simple threshold', image_thresh)
cv2.imshow('adaptive threshold', image_adapt)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Edge Detection

image = cv2.imread('test_image.jpg')

# Canny edge detection
image_edge = cv2.Canny(image, 350, 600)
    # 3 types of edge detectors: Sobel, Laplace, Canny

cv2.imshow('normal', image)
cv2.imshow('Canny', image_edge)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Edge Detection: Dialating and Eroding

# dialating makes all the borders thicker, bigger #s in np.ones() -> thicker lines
image_edge_d = cv2.dilate(image_edge, np.ones((2, 2), dtype=np.int8))

# eroding does the opposite of dialting, makes lines thinner
image_edge_e = cv2.erode(image_edge, np.ones((2, 2), dtype=np.int8))

cv2.imshow('normal', image)
cv2.imshow('edge detection', image_edge)
cv2.imshow('edge detection dialated', image_edge_d)
cv2.imshow('edge detection eroded', image_edge_e)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drawing
    # draw Lines, Rectangles, Circles, or Text

image = cv2.imread('test_image.jpg')

# Line
print(image.shape)
cv2.line(image, (100, 100), (500, 200), (0, 255, 0), 3)
    # parameters: image, (x1, y1), (x2, y2), color, thickness
    # the line starts and ends at (x1, y1) and (x2, y2)

cv2.imshow('window', image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drawing: Rectangle
print(image.shape)
cv2.rectangle(image, (150, 50), (400, 150), (255, 0, 0), 6)
    # parameters: image, (x1, y1), (x2, y2), color, thickness
    # (x1, y1) is the upper left point, (x2, y2) is bottom right

cv2.imshow('window', image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drawing: Circle
print(image.shape)
cv2.circle(image, (180, 280), 100, (0, 0, 255), -1)
    # parameters: image, (x, y), radius, color, thickness
    # (x, y) is the center of the circle
    # thickness of -1 fills in the shape
    # shapes are allowed to go off the bounds of the image

cv2.imshow('window', image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drawing: Text
print(image.shape)
cv2.putText(image, 'hello world!', (250, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 25, 100), 4)
    # parameters: image, text, (x, y), font, size, color, thickness
    # (x, y) is the bottom left corner of the text

cv2.imshow('window', image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Contours: detecting groups of pixels

image = cv2.imread('birds.png')
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
image_contour = image.copy() # we will draw contours on this image later

# threshold that takes a threshold and inverts the image afterward
ret, image_thresh = cv2.threshold(image_gray, 127, 255, cv2.THRESH_BINARY_INV)
    # need black and white image for contours
    # inverses the colors because we will be detecting the white pixels

contours, hierarchy = cv2.findContours(image_thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    # 'contours' will be a list where each element is an isolated object (isolated white shapes surrounded by black pixels in the image) 
for cnt in contours:
    if cv2.contourArea(cnt) > 200:  # only continues if size of the object is large enough (>200 pixels)
        cv2.drawContours(image_contour, cnt, -1, (0, 0, 255), 2)
        # parameters: image, contour, -1, color, thickness
        # draws contours on the original image

cv2.imshow('image', image)
cv2.imshow('threshold', image_thresh)
cv2.imshow('contours', image_contour)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Drawing Bounding Boxes around each object
for cnt in contours:
    if cv2.contourArea(cnt) > 200:  # only continues if size of the object is large enough (>200 pixels)
        x1, y1, w, h = cv2.boundingRect(cnt) # finds a bounding box for each object
        cv2.rectangle(image, (x1, y1), (x1 + w, y1 + h), (0, 0, 255), 2)

cv2.imshow('image', image)
cv2.imshow('threshold', image_thresh)
cv2.imshow('contours', image_contour)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
